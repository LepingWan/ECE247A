# Problem 1

## (a)

Let $Q$ be a real orthogonal matrix. Since $Q$ is orthogonal,
$
Q^TQ = QQ^T = I.
$

### (i)
Squaring norm gives $\|x\|_2^2 = \sum_{j=1}^n x_j^2$. Also, multiplying the row vector $x^T$ by the column vector $x$ gives
$$
x^Tx = \sum_{j=1}^n x_j x_j = \sum_{j=1}^n x_j^2.
$$
Therefore, $\|x\|_2^2 = x^Tx$.

Then
$
\|Qx\|_2^2
= (Qx)^T(Qx)
= x^TQ^TQx.
$

Since $Q^TQ=I$,
$
\|Qx\|_2^2
= x^TIx
= x^Tx
= \|x\|_2^2.
$

Since norms are nonnegative, taking square roots gives
$
\|Qx\|_2 = \|x\|_2.
$


### (ii)

The inner product is defined by
$
\langle x,y\rangle = x^Ty.
$

Then
$
\langle Qx,Qy\rangle
= (Qx)^T(Qy)
= x^TQ^TQy.
$

Again using $Q^TQ=I$,

$$
\langle Qx,Qy\rangle
= x^Ty
= \langle x,y\rangle.
$$

Therefore,

$$
\langle Qx,Qy\rangle = \langle x,y\rangle.
$$


### (iii)

Suppose $\lambda \in \mathbb{C}$ is an eigenvalue of $Q$. Then there is some nonzero
$v\in\mathbb{C}^n$ such that

$$
Qv = \lambda v.
$$

Taking the complex conjugate of both sides gives

$$
\overline{Qv} = \overline{\lambda v}.
$$

Since $Q$ is a real matrix, $\overline{Q}=Q$. Therefore,

$$
Q\overline{v} = \overline{\lambda}\,\overline{v}.
$$

Since $v\neq 0$, we also have $\overline{v}\neq 0$. Thus $\overline{v}$ is an
eigenvector of $Q$ with eigenvalue $\overline{\lambda}$.

Therefore,

$$
\overline{\lambda}\text{ is also an eigenvalue of }Q.
$$


### (iv)

Suppose $\{u_1,\ldots,u_k\}$ is an orthonormal set. This means

$$
u_i^Tu_j =
\begin{cases}
1, & i=j,\\
0, & i\neq j.
\end{cases}
$$

Consider the vectors

$$
Qu_1,\ldots,Qu_k.
$$

Using part (ii),

$$
\langle Qu_i,Qu_j\rangle
= \langle u_i,u_j\rangle.
$$

Therefore,

$$
(Qu_i)^T(Qu_j)
=
\begin{cases}
1, & i=j,\\
0, & i\neq j.
\end{cases}
$$

So $\{Qu_1,\ldots,Qu_k\}$ is also an orthonormal set.

Thus,

$$
Q\text{ maps orthonormal sets to orthonormal sets}.
$$


## (b)

Let $A\in\mathbb{R}^{m\times n}$.

### (i)

First consider $A^TA$. Its transpose is

$$
(A^TA)^T = A^T(A^T)^T = A^TA.
$$

Therefore $A^TA$ is symmetric.

Similarly,

$$
(AA^T)^T = (A^T)^TA^T = AA^T,
$$

so $AA^T$ is also symmetric.

To show that $A^TA$ is positive semidefinite, let $x\in\mathbb{R}^n$. Then

$$
x^TA^TAx
= (Ax)^T(Ax)
= \|Ax\|_2^2
\geq 0.
$$

Therefore $A^TA$ is positive semidefinite.

Similarly, for any $y\in\mathbb{R}^m$,

$$
y^TAA^Ty
= (A^Ty)^T(A^Ty)
= \|A^Ty\|_2^2
\geq 0.
$$

Therefore,

$$
A^TA\text{ and }AA^T\text{ are both symmetric and positive semidefinite}.
$$


### (ii)

Suppose $v$ is an eigenvector of $A^TA$ with eigenvalue $\lambda\neq 0$. Then

$$
A^TAv = \lambda v.
$$

First, $Av$ cannot be zero. If $Av=0$, then

$$
A^TAv=0,
$$

which would imply $\lambda v=0$. Since $\lambda\neq0$ and $v\neq0$, this is
impossible.

Now multiply the eigenvalue equation by $A$:

$$
A(A^TAv) = A(\lambda v).
$$

Therefore,

$$
AA^T(Av) = \lambda(Av).
$$

Since $Av\neq0$, this shows that $\lambda$ is also an eigenvalue of $AA^T$.

The same argument works in the other direction. If

$$
AA^Tu = \lambda u,
$$

where $\lambda\neq0$, then $A^Tu\neq0$, and

$$
A^TA(A^Tu) = \lambda(A^Tu).
$$

Thus every nonzero eigenvalue of one matrix is also a nonzero eigenvalue of
the other.

Their multiplicities also agree: for each $\lambda\neq0$, the map $v\mapsto Av$ between the corresponding eigenspaces has inverse $u\mapsto \lambda^{-1}A^Tu$. Thus these eigenspaces have the same dimension. Since both matrices are real symmetric, they are diagonalizable, so the algebraic multiplicities also agree.

Therefore,

$$
A^TA\text{ and }AA^T\text{ have the same nonzero eigenvalues}.
$$

## (a)

### (i)

$$
P(T=n)=(1-p)^{n-1}p.
$$

### (ii)

At least $n$ items are inspected if and only if the first $n-1$ items are non-defective. Item $n$ can be either defective or non-defective. Therefore,

$$
P(T\geq n)=(1-p)^{n-1}.
$$

### (iii)

Exactly $k$ non-defective items are inspected before stopping if the first $k$ items are non-defective and the next item is defective. so,

$$
P(T=k+1)=(1-p)^k p.
$$

### (iv)

Let $E$ be the event that the first $n-1$ items are all non-defective. The event $T=n$ implies $E$, so

$$
P(E\mid T=n)=\frac{P(E\cap\{T=n\})}{P(T=n)}
=\frac{P(T=n)}{P(T=n)}=1.
$$

### (v)

Given that at least $n$ items are inspected, the first $n-1$ items are non-defective. By independence, this does not change the probability that item $n$ is defective. Equivalently,

$$
P(\text{item }n\text{ is defective}\mid T\geq n)
=\frac{P(T=n)}{P(T\geq n)}
=\frac{(1-p)^{n-1}p}{(1-p)^{n-1}}=p.
$$

## (b)

Suppose $P(X=+1)=\pi$, $P(X=-1)=1-\pi$, and $Y=X+N$, where $N\sim\mathcal{N}(0,\sigma^2)$ is independent of $X$. Assume $0<\pi<1$ and $\sigma>0$.

### (i)

Conditioning on $X$ fixes its value, so adding the Gaussian noise gives

$$
Y\mid X=+1\sim\mathcal{N}(1,\sigma^2),
\qquad
Y\mid X=-1\sim\mathcal{N}(-1,\sigma^2).
$$

The conditional PDFs are therefore

$$
f_{Y\mid X}(y\mid X=+1)
=\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(-\frac{(y-1)^2}{2\sigma^2}\right),
$$

$$
f_{Y\mid X}(y\mid X=-1)
=\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(-\frac{(y+1)^2}{2\sigma^2}\right).
$$

Here, $\pi$ in the Gaussian normalization constant is the usual mathematical constant; $\pi$ in $P(X=+1)=\pi$ is the prior probability.

### (ii)

By Bayes' rule,

$$
P(X=+1\mid Y=y)
=\frac{\pi f_{Y\mid X}(y\mid X=+1)}
{\pi f_{Y\mid X}(y\mid X=+1)+(1-\pi)f_{Y\mid X}(y\mid X=-1)}.
$$

The common Gaussian normalization constant cancels. Also,

$$
\frac{f_{Y\mid X}(y\mid X=-1)}{f_{Y\mid X}(y\mid X=+1)}
=\exp\left(\frac{(y-1)^2-(y+1)^2}{2\sigma^2}\right)
=\exp\left(-\frac{2y}{\sigma^2}\right).
$$

Therefore,

$$
P(X=+1\mid Y=y)
=\frac{1}{1+\frac{1-\pi}{\pi}\exp\left(-\frac{2y}{\sigma^2}\right)}.
$$

### (iii)

The posterior probability is at least $0.5$ exactly when the weighted likelihood for $X=+1$ is at least that for $X=-1$:

$$
\pi f_{Y\mid X}(y\mid X=+1)
\geq(1-\pi)f_{Y\mid X}(y\mid X=-1).
$$

Dividing by the positive quantity $\pi f_{Y\mid X}(y\mid X=-1)$ gives

$$
\frac{f_{Y\mid X}(y\mid X=+1)}{f_{Y\mid X}(y\mid X=-1)}
=\exp\left(\frac{2y}{\sigma^2}\right)
\geq\frac{1-\pi}{\pi}.
$$

Taking natural logarithms and multiplying by $\sigma^2/2>0$,

$$
y\geq\frac{\sigma^2}{2}\ln\left(\frac{1-\pi}{\pi}\right).
$$

Thus the threshold is

$$
\gamma^\star=\frac{\sigma^2}{2}\ln\left(\frac{1-\pi}{\pi}\right),
$$

and $P(X=+1\mid Y=y)\geq0.5$ if and only if $y\geq\gamma^\star$. For equal priors ($\pi=1/2$), the threshold is $0$. If $\pi=0$, the detector always chooses $-1$; if $\pi=1$, it always chooses $+1$.

# Problem 3

Gradients with respect to vectors are column vectors. For a matrix $A$, $(\nabla_A f)_{ij}=\partial f/\partial A_{ij}$, so the gradient has the same dimensions as $A$.

## (a)

For $f=x^TAy$, treating $A$ and $y$ as constants,

$$
\nabla_x f=Ay\in\mathbb{R}^n.
$$

## (b)

Writing the scalar as $f=y^TA^Tx$,

$$
\nabla_y f=A^Tx\in\mathbb{R}^m.
$$

## (c)

Since $f=x^TAy=\sum_{i=1}^n\sum_{j=1}^m x_i A_{ij}y_j$, we have $\partial f/\partial A_{ij}=x_i y_j$. Therefore,

$$
\nabla_A f=xy^T\in\mathbb{R}^{n\times m}.
$$

## (d)

For $f=x^TAx+b^Tx$, the quadratic term contributes $(A+A^T)x$ and the linear term contributes $b$. Thus,

$$
\nabla_x f=(A+A^T)x+b.
$$

No symmetry of $A$ is assumed. If $A$ is symmetric, this simplifies to $2Ax+b$.

## (e)

Using $\operatorname{tr}(AB)=\sum_{i,j}A_{ij}B_{ji}$,

$$
\nabla_A f=B^T.
$$

## (f)

For $f=\operatorname{tr}(BA+A^TB+A^2B)$, the first two terms give

$$
\nabla_A\operatorname{tr}(BA)=B^T,
\qquad
\nabla_A\operatorname{tr}(A^TB)=B.
$$

Here $A^2=AA$ denotes matrix multiplication. For the last term, apply the product rule and the cyclic property of the trace:

$$
d\operatorname{tr}(A^2B)
=\operatorname{tr}\big((dA)AB+A(dA)B\big)
=\operatorname{tr}\big((AB+BA)dA\big).
$$

Comparing with $df=\operatorname{tr}((\nabla_A f)^T dA)$, this term contributes $(AB+BA)^T$. Adding all three contributions gives

$$
\nabla_A f=B^T+B+(AB+BA)^T
=B^T+B+B^TA^T+A^TB^T.
$$

## (g)

Treating $B$ and $\lambda$ as constants, write the squared Frobenius norm as

$$
f=\|A+\lambda B\|_F^2
=\sum_{i,j}(A_{ij}+\lambda B_{ij})^2.
$$

Differentiating each entry gives $\partial f/\partial A_{ij}=2(A_{ij}+\lambda B_{ij})$. Therefore,

$$
\nabla_A f=2(A+\lambda B).
$$

# Problem 4

Let each input $x^{(i)}\in\mathbb{R}^d$ and output $y^{(i)}\in\mathbb{R}^r$, so $W\in\mathbb{R}^{r\times d}$. We minimize

$$
L(W)=\frac12\sum_{i=1}^n\|y^{(i)}-Wx^{(i)}\|_2^2.
$$

Collect the training examples as columns:

$$
X=\begin{bmatrix}x^{(1)}&\cdots&x^{(n)}\end{bmatrix}\in\mathbb{R}^{d\times n},
\qquad
Y=\begin{bmatrix}y^{(1)}&\cdots&y^{(n)}\end{bmatrix}\in\mathbb{R}^{r\times n}.
$$

Then the objective can be written using the Frobenius norm and trace:

$$
L(W)=\frac12\|Y-WX\|_F^2
=\frac12\operatorname{tr}\big((Y-WX)(Y-WX)^T\big).
$$

Expanding and using the fact that the two cross terms have the same trace gives

$$
L(W)=\frac12\operatorname{tr}(YY^T)
-\operatorname{tr}(WXY^T)
+\frac12\operatorname{tr}(WXX^TW^T).
$$

The first term is constant with respect to $W$. Using the provided derivatives and the symmetry of $XX^T$,

$$
\nabla_W L
=-(XY^T)^T+\frac12\big(WXX^T+W(XX^T)^T\big)
=-YX^T+WXX^T.
$$

Setting the gradient to zero gives the normal equation

$$
WXX^T=YX^T.
$$

If $XX^T$ is invertible (equivalently, $X$ has full row rank), multiplying on the right by its inverse gives the unique solution

$$
W^\star=YX^T(XX^T)^{-1}
=\left(\sum_{i=1}^n y^{(i)}(x^{(i)})^T\right)
\left(\sum_{i=1}^n x^{(i)}(x^{(i)})^T\right)^{-1}.
$$

This stationary point is a global minimum because $L(W)$ is convex. In particular, for any perturbation $H$, its second directional derivative is $\|HX\|_F^2\geq0$. When $X$ has full row rank, this is positive for every nonzero $H$, so the minimum is unique.

If $XX^T$ is singular, an ordinary inverse does not exist and the minimizer is not unique. The minimum-Frobenius-norm solution is

$$
W^\star=YX^\dagger=YX^T(XX^T)^\dagger,
$$

where $\dagger$ denotes the Moore-Penrose pseudoinverse. This solution also satisfies the normal equation.

# Problem 5

The gradient of loss with respect to $\theta$ gives

$$
\nabla_\theta J
=-\sum_{i=1}^N\hat{x}^{(i)}\left(y^{(i)}-\theta^T\hat{x}^{(i)}\right)+\lambda\theta
=\left(\sum_{i=1}^N\hat{x}^{(i)}(\hat{x}^{(i)})^T+\lambda I_d\right)\theta
-\sum_{i=1}^N\hat{x}^{(i)}y^{(i)}.
$$

Setting the gradient to zero gives

$$
\left(\sum_{i=1}^N\hat{x}^{(i)}(\hat{x}^{(i)})^T+\lambda I_d\right)\theta^\star
=\sum_{i=1}^N\hat{x}^{(i)}y^{(i)}.
$$

Thus, whenever the coefficient matrix is invertible,

$$
\theta^\star
=\left(\sum_{i=1}^N\hat{x}^{(i)}(\hat{x}^{(i)})^T+\lambda I_d\right)^{-1}
\sum_{i=1}^N\hat{x}^{(i)}y^{(i)}.
$$

Equivalently, define the design matrix with training inputs as rows and the target vector as

$$
\hat{X}=\begin{bmatrix}(\hat{x}^{(1)})^T\\\vdots\\(\hat{x}^{(N)})^T\end{bmatrix}\in\mathbb{R}^{N\times d},
\qquad
y=\begin{bmatrix}y^{(1)}\\\vdots\\y^{(N)}\end{bmatrix}\in\mathbb{R}^N.
$$

Then

$$
\theta^\star=(\hat{X}^T\hat{X}+\lambda I_d)^{-1}\hat{X}^Ty.
$$

For $\lambda>0$, the Hessian is positive definite: for every nonzero $z\in\mathbb{R}^d$,

$$
z^T\nabla_\theta^2J\,z
=z^T(\hat{X}^T\hat{X}+\lambda I_d)z
=\|\hat{X}z\|_2^2+\lambda\|z\|_2^2>0.
$$

Therefore the inverse exists and $\theta^\star$ is the unique global minimizer, even if $\hat{X}$ does not have full column rank.

When $\lambda=0$, this reduces to ordinary least squares. If $\hat{X}^T\hat{X}$ is singular, the minimum-norm solution is $\theta^\star=\hat{X}^\dagger y$.

The given penalty applies to every component of $\theta$, including an intercept coefficient if one is included in $\hat{x}^{(i)}$.

### Fitting data to the model (5 points)

Here, we'll now do regression to polynomial models of orders 1 to 5.  Note, the order 1 model is the linear model you prior fit.

```python
N = 5
xhats = []
thetas = []

# ==================== #
# START YOUR CODE HERE #
# ==================== #

# GOAL: create a variable thetas.
# thetas is a list, where theta[i] are the model parameters for the polynomial fit of order i+1.
#   i.e., thetas[0] is equivalent to theta above.
#   i.e., thetas[1] should be a length 3 np.array with the coefficients of the x^2, x, and 1 respectively.
#   ... etc.

for degree in range(1, N + 1):
    xhat = np.vander(x, N=degree + 1).T
    xhats.append(xhat)
    thetas.append(np.linalg.lstsq(xhat.T, y, rcond=None)[0])

# ================== #
# END YOUR CODE HERE #
# ================== #
```

```python
# Plot the data
f = plt.figure()
ax = f.gca()
ax.plot(x, y, ".")
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")

# Plot the regression lines
plot_xs = []
for i in np.arange(N):
    if i == 0:
        plot_x = np.vstack((np.linspace(min(x), max(x), 50), np.ones(50)))
    else:
        plot_x = np.vstack((plot_x[-2] ** (i + 1), plot_x))
    plot_xs.append(plot_x)

for i in np.arange(N):
    ax.plot(plot_xs[i][-2, :], thetas[i].dot(plot_xs[i]))

labels = ["data"]
[labels.append("n={}".format(i + 1)) for i in np.arange(N)]
bbox_to_anchor = (1.3, 1)
lgd = ax.legend(labels, bbox_to_anchor=bbox_to_anchor)
plt.show()
```

### Calculating the training error (5 points)

Here, we'll now calculate the training error of polynomial models of orders 1 to 5.

```python
training_errors = []

# ==================== #
# START YOUR CODE HERE #
# ==================== #

# GOAL: create a variable training_errors, a list of 5 elements,
# where training_errors[i] are the training loss for the polynomial fit of order i+1.
# Use mean squared error for both training and testing loss.
for xhat, coefficients in zip(xhats, thetas):
    training_errors.append(float(np.mean((coefficients.dot(xhat) - y) ** 2)))

# ================== #
# END YOUR CODE HERE #
# ================== #

print("Training errors are: \n", training_errors)
```

### QUESTIONS

(1) What polynomial has the best training error?

(2) Why is this expected?

### ANSWERS

(1) The degree-5 polynomial has the smallest training mean squared error, approximately $0.002267$, Although degree 3 and 4 are very close.

(2) This is expected because higher-degree polynomials have more flexibility to fit the training data. They can match the overall curve and also adjust to some of the random noise, giving a smaller training error. However, fitting the training data better does not necessarily mean they will predict new data better.
